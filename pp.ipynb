```python
#Identify unusually long deliveries using IQR.?

# Calculate Q1 and Q3
Q1 = df["Delivery_Time_Min"].quantile(0.25)
Q3 = df["Delivery_Time_Min"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate upper limit
upper_limit = Q3 + 1.5 * IQR

# Find unusually long deliveries
outliers = df[df["Delivery_Time_Min"] > upper_limit]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Upper Limit:", upper_limit)

print("Outliers:")
print(outliers[["Order_ID", "Delivery_Time_Min"]])

print("Total Outliers:", len(outliers))
```

```python
#Create delivery-time bands such as Fast, Standard and Delayed?

# Create delivery time groups
df["Delivery_Band"] = pd.cut(
    df["Delivery_Time_Min"],
    bins=[0, 30, 45, 1000],
    labels=["Fast", "Standard", "Delayed"]
)

# Count orders in each group
print(df["Delivery_Band"].value_counts())
```
